```python
import bricks2marble as b2m
```

```python
fasta_path = ""
```

```python
fasta = b2m.load_fasta(
    fasta_path,
    T=10_000,
    restrict=-1,
)
```

```python
print(f"{fasta.T = }  -> {fasta.nuc.shape = }")
print(f"{fasta = !s}")
print(f"{fasta.nuc.size = }")

print("\nreshape into smaller chunks...\n")
# ---------------------------
fasta = fasta.resample(5_000)
# ---------------------------

print(f"{fasta.T = }  -> {fasta.nuc.shape = }")
print(f"{fasta = !s}")
print(f"{fasta.nuc.size = }")
```

```python
seq = fasta[0].positions(1000, 15532)
# ----------------------------------
print(f"{seq.nuc.shape = }")
print(f"{seq = }")
print(f"{seq.nuc.size = }")
```

```python
seq = seq.positions(1200, 6200)
# ---------------------------------

print(f"{seq.nuc.shape = }")
print(f"{seq = }")
print(f"{seq.nuc.size = }")
```
