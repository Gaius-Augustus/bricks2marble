```python
import bricks2marble as b2m
```

```python
fasta_path = ""
```

```python
fasta = b2m.load_fasta(
    fasta_path,
    T=10_000,
    restrict=1_000_000,
    repeat_masking=True,
)
```

```python
print(f"{fasta.T = }  -> {fasta.nuc.shape = }")
print(f"{fasta.comprises() = }")
print(f"{fasta.nuc.size = }")

print("\nreshape into smaller chunks...\n")
# ---------------------------
fasta = fasta.resample(5_000)
# ---------------------------

print(f"{fasta.T = }  -> {fasta.nuc.shape = }")
print(f"{fasta.comprises() = }")
print(f"{fasta.nuc.size = }")
```

```python
fasta = fasta.positions(1000, 15532)
# ----------------------------------
print(f"{fasta.nuc.shape = }")
print(f"{fasta.comprises() = }")
print(f"{fasta.nuc.size = }")
```

```python
fasta = fasta.positions(1000, 2000)
# ---------------------------------

print(f"{fasta.nuc.shape = }")
print(f"{fasta.comprises() = }")
print(f"{fasta.nuc.size = }")
```
